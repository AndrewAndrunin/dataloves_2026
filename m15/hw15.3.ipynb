# Тести для малих вибірок та пропорцій



**Завдання 1**. E-commerce компанія після редизайну сайту підозрює, що **середній час до покупки (time-to-purchase)** користувачів **збільшився**.
Історично середній час від першого заходу на сайт до покупки становив $\mu_0 = 123.7$ хвилин.

Команда аналітиків випадково обрала дані **7 покупців після редизайну**:
`128, 135, 121, 142, 126, 151, 123`

З допомогою наявних даних зʼясуйте, чи збільшився середній час покупки після редизайну.

Для цього виконайте наступні 6 кроків. Правильне виконання кожного з кроків оцінюється в 1 бал.

1. Запишіть нульову та альтернативну гіпотези і визначте тип тесту.
2. Обчисліть вибіркові статистики: $\bar x$, $s$, $n$.
3. Оберіть тип тесту та виконайте його (знайдіть критичне значення тесту, статистику тесту та р-значення) будь-яким пасуючим способом, щоб перевірити гіпотезу на рівні значущості $\alpha = 0.10$.
4. Прийміть рішення, чи відхиляєте ви гіпотезу $H_0$ використовуючи p-value.
5. Напишіть висновок: чи справді редизайн сайту подовжив час до покупки?
6. Чи зміниться ваше рішення при зміні рівня значущості на $\alpha = 0.05$.


**Крок 1. Гіпотези і тип тесту**

Компанія підозрює, що час до покупки **збільшився**.

- $H_0: \mu = 123.7$ хв — час до покупки не змінився;
- $H_a: \mu > 123.7$ хв — час до покупки збільшився.

Це **правосторонній** тест.

```python
import numpy as np
from scipy import stats

times = np.array([128, 135, 121, 142, 126, 151, 123])
mu0 = 123.7

x_bar = times.mean()
s = times.std(ddof=1)
n = len(times)

print(f'x̄ = {x_bar:.2f} хв')
print(f's = {s:.2f} хв')
print(f'n = {n}')
```

**Крок 3. Вибір тесту**

Вибірка мала ($n = 7 < 30$), а стандартне відхилення популяції $\sigma$ невідоме. Тому використовуємо **одновибірковий t-тест** (правосторонній) з $df = n - 1 = 6$ ступенями свободи.

```python
alpha = 0.10
dof = n - 1

SE = s / np.sqrt(n)
t_stat = (x_bar - mu0) / SE
t_crit = stats.t.ppf(1 - alpha, dof)
p_value = 1 - stats.t.cdf(t_stat, dof)

print(f't-статистика = {t_stat:.4f}')
print(f't критичне   = {t_crit:.4f}  (α = {alpha}, df = {dof})')
print(f'p-value      = {p_value:.4f}')

res = stats.ttest_1samp(times, popmean=mu0, alternative='greater')
print(f'scipy: t = {res.statistic:.4f}, p-value = {res.pvalue:.4f}')
```

```python
if p_value <= alpha:
    print(f'p-value = {p_value:.3f} ≤ α = {alpha} → відхиляємо H0')
else:
    print(f'p-value = {p_value:.3f} > α = {alpha} → не відхиляємо H0')
```

**Крок 4. Рішення.** $p\text{-value} \approx 0.042 < \alpha = 0.10$, тому **відхиляємо $H_0$**. Те саме видно з критичного значення: $t \approx 2.07 > t_{кр} \approx 1.44$.

**Крок 5. Висновок.** На рівні значущості 10 % дані підтверджують, що після редизайну середній час до покупки збільшився. У вибірці він становить 132.3 хв проти 123.7 хв раніше.

Зауваження: вибірка дуже мала (7 покупців). t-тест припускає, що час до покупки розподілений приблизно нормально. Тому висновок варто перевірити на більшій вибірці.

```python
alpha_05 = 0.05
t_crit_05 = stats.t.ppf(1 - alpha_05, dof)

print(f't критичне  = {t_crit_05:.4f}  (α = {alpha_05})')
print(f't-статистика = {t_stat:.4f}, p-value = {p_value:.4f}')
print('Рішення:', 'відхиляємо H0' if p_value <= alpha_05 else 'не відхиляємо H0')
```

**Крок 6.** При $\alpha = 0.05$ рішення **не змінюється**: $p\text{-value} \approx 0.042 < 0.05$, а $t \approx 2.07 > t_{кр} \approx 1.94$. Ми й далі відхиляємо $H_0$.

Але запас малий. Рішення змінилося б лише при $\alpha < 0.042$, наприклад при $\alpha = 0.01$.

**Завдання 2.**

До спеціальної рекламної кампанії **23%** дорослих упізнавали логотип компанії. Після завершення кампанії відділ маркетингу провів опитування: з **1200** випадково відібраних дорослих **311** упізнали логотип.

Перевірте на рівні значущості **3%** ($\alpha=0.03$), чи дають ці дані достатні підстави стверджувати, що **тепер більше ніж 23%** дорослих упізнають логотип компанії. Для розвʼязку використовуйте бібліотеку `statsmodels`.

Зробіть висновок, чи зросла впізнаваність логотипу.

Додатково, обчисліть довірчий інтревал на заданому рівні значущості і проінтерпретуйте текстом - як він додатково пояснює прийняте нами рішення?

**Гіпотези**

Позначимо $p$ — частку дорослих, які впізнають логотип після кампанії.

- $H_0: p = 0.23$ — впізнаваність не змінилася;
- $H_a: p > 0.23$ — впізнаваність зросла.

Це **правосторонній** z-тест для однієї пропорції, $\alpha = 0.03$.

```python
from statsmodels.stats.proportion import proportions_ztest, proportion_confint

count = 311
nobs = 1200
p0 = 0.23
alpha = 0.03

p_hat = count / nobs
z_stat, p_value = proportions_ztest(count, nobs, value=p0, alternative='larger')

print(f'Вибіркова частка p̂ = {p_hat:.4f}')
print(f'z-статистика = {z_stat:.4f}')
print(f'p-value = {p_value:.4f}')
print('Рішення:', 'відхиляємо H0' if p_value <= alpha else 'не відхиляємо H0')

ci_low, ci_high = proportion_confint(count, nobs, alpha=alpha, method='normal')
print(f'97% довірчий інтервал для p: [{ci_low:.4f}, {ci_high:.4f}]')

ci_low_one_sided = proportion_confint(count, nobs, alpha=2 * alpha, method='normal')[0]
print(f'Односторонній 97% інтервал: p ≥ {ci_low_one_sided:.4f}')
```

**Висновок**

- Після кампанії логотип упізнали 311 з 1200 дорослих, тобто $\hat p \approx 25.9\%$.
- $z \approx 2.31$, $p\text{-value} \approx 0.011 < \alpha = 0.03$. **Відхиляємо $H_0$.**
- Отже, дані дають достатні підстави стверджувати, що тепер логотип упізнають **більше ніж 23 %** дорослих. Впізнаваність зросла.

**Як це пояснює довірчий інтервал**

- 97% довірчий інтервал для частки — приблизно **[23.2 %, 28.7 %]**.
- Увесь інтервал лежить **вище за 23 %**. Значення з $H_0$ ($p = 0.23$) не входить до правдоподібних значень частки. Це узгоджується з рішенням відхилити $H_0$.
- Для одностороннього тесту точніше дивитися на односторонню межу: $p \ge 23.5\%$. Вона теж вища за 23 %.
- Нижня межа близька до 23 %. Отже, зростання є, але воно може бути невеликим: приблизно від 0.2 до 5.7 відсоткового пункту.
